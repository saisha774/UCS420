```python
#que 3
import random
random.seed(1024170061)
```

```python
numbers=[random.randint(100,900) for i in range(100)]
print(numbers)
```

```python
odd_num=[i for i in numbers if i%2!=0]
print(odd_num)
print("count is:", len(odd_num))
```

```python
even_num=[i for i in numbers if i%2==0]
print(even_num)
print("count is:", len(even_num))
```

```python
def prime(i):
    isprime=True
    if i<2:
        isprime=False
        return isprime
    for j in range(2,(i//2)+1,1):
        if i%j==0:
            isprime=False
    return isprime
prime_number=[k for k in numbers if prime(k)]
print(prime_number)
print("count is:",len(prime_number))
```

```python
most_frequent = max(set(numbers), key=numbers.count)
print("Most frequent number:", most_frequent)
print("Frequency:", numbers.count(most_frequent))
```
